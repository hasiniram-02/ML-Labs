```python
# Task: Exploratory Data Analysis (EDA) on Titanic Dataset

# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load Titanic dataset directly from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Display first five rows
print("First Five Rows:")
print(df.head())

# Display last five rows
print("\nLast Five Rows:")
print(df.tail())

# Shape of dataset
print("\nShape of Dataset:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Dataset information
print("\nDataset Information:")
print(df.info())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Survival count
print("\nSurvival Count:")
print(df["Survived"].value_counts())

# Gender count
print("\nGender Count:")
print(df["Sex"].value_counts())

# Passenger class count
print("\nPassenger Class Count:")
print(df["Pclass"].value_counts())

# Correlation (only numeric columns)
print("\nCorrelation Matrix:")
print(df.corr(numeric_only=True))

# -------------------------------
# Data Visualization
# -------------------------------

# Survival Count
plt.figure(figsize=(5,4))
df["Survived"].value_counts().plot(kind="bar")
plt.title("Survival Count-24eu02024")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()

# Gender Distribution
plt.figure(figsize=(5,4))
df["Sex"].value_counts().plot(kind="bar")
plt.title("Gender Distribution-24eu02024")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# Passenger Class Distribution
plt.figure(figsize=(5,4))
df["Pclass"].value_counts().plot(kind="bar")
plt.title("Passenger Class Distribution-24eu02024")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()

# Age Distribution
plt.figure(figsize=(6,4))
df["Age"].plot(kind="hist", bins=20)
plt.title("Age Distribution-24eu02024")
plt.xlabel("Age")
plt.show()

# Fare Distribution
plt.figure(figsize=(6,4))
df["Fare"].plot(kind="hist", bins=20)
plt.title("Fare Distribution-24eu02024")
plt.xlabel("Fare")
plt.show()

# Boxplot of Age
plt.figure(figsize=(5,4))
df.boxplot(column="Age")
plt.title("Boxplot of Age-24eu02024")
plt.show()
```
